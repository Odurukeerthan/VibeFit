```python
import cv2
import numpy as np
```

```python
img=cv2.imread(r"C:\Users\obunn\OneDrive\Pictures\desktop-367124.jpg")
```

```python
cv2.imshow("original",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
width = 800
aspect_ratio = img.shape[1] / img.shape[0] 
height = int(width / aspect_ratio)
image = cv2.resize(img, (width, height))
cv2.imshow("Resized", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
b,g,r=cv2.split(image)
cv2.imshow("blue",b)
cv2.imshow("green",g)
cv2.imshow("red",r)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_image=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
cv2.imshow("Gray", gray_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
matrix=np.array(b)
b_t=matrix.T
print(b_t)
print(b)
```

```python
cv2.imshow("blue",b)
cv2.imshow("bluw_t",b_t)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
